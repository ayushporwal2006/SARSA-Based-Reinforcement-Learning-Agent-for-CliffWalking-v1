```python
import gymnasium as gym
import numpy as np
import random
```

```python
alpha = 0.5
gamma = 0.99
episodes = 500
epsilon = 0.1
```

```python
Q = np.zeros((48,4))
Q
```

```python
def epsilon_greedy(state):
    if random.random() < epsilon:
        return env.action_space.sample()
    else:
        return np.argmax(Q[state])
```

```python
for episode in range(episodes):
    render = (episode %50 == 0)
    if render :
        env = gym.make("CliffWalking-v1", render_mode ='human')
        print(f"rendering at {episode+1}")
    else:
        env = gym.make("CliffWalking-v1")
    done = False
    state, _ = env.reset()
    
    total_rewards = 0
    episode_len = 0

    while not done:
        action = epsilon_greedy(state)    
        next_state, reward, terminated, truncated , _ = env.step(action)
        done = terminated or  truncated
        # Q learning updation condition
        Q[state, action] += alpha * (reward + gamma* np.max( Q[next_state] -  Q[state, action]))

        state = next_state
        episode_len +=1
        total_rewards += reward

    print(f"Episode = {episode} : total_rewards = {total_rewards}, episode_length = {episode_len}")
    env.close()
```

```python
# What did our agent learn

env = gym.make("CliffWalking-v1", render_mode ='human')
state ,_ = env.reset()
total_rewards = 0
episode_len = 0
done = False

while not done:
    action = np.argmax(Q[state])
    next_state , reward, terminated, truncated , _ = env.step(action)
    done = terminated or  truncated
    state = next_state
    episode_len +=1
    total_rewards += reward
print(f"total_reward = {total_rewards} & episode_length = {episode_len}")
env.close()
```
